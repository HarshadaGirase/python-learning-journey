### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
# Get CTC from user
ctc = float(input("Enter your CTC in rupees: "))

# Fixed percentages as per problem
hra_percent = 10  # 10%
da_percent = 5    # 5%
pf_percent = 3    # 3%

# Calculate deductions
hra = (hra_percent / 100) * ctc
da = (da_percent / 100) * ctc
pf = (pf_percent / 100) * ctc

# Calculate taxable salary (CTC minus all deductions except tax)
taxable_salary = ctc - (hra + da + pf)

# Calculate tax based on slabs (in lakhs)
if taxable_salary <= 500000:
    tax_percent = 0
elif taxable_salary <= 1000000:  # Between 5-10 lakhs
    tax_percent = 10
elif taxable_salary <= 2000000:  # Between 10-20 lakhs
    tax_percent = 20
else:  # Above 20 lakhs
    tax_percent = 30

# Calculate tax amount
tax_amount = (tax_percent / 100) * taxable_salary

# Calculate total deductions
total_deductions = hra + da + pf + tax_amount

# Calculate in-hand salary
in_hand_salary = ctc - total_deductions

# Display results
print("\n" + "="*50)
print("SALARY BREAKDOWN")
print("="*50)
print(f"CTC: ₹{ctc:,.2f}")
print(f"HRA (10%): ₹{hra:,.2f}")
print(f"DA (5%): ₹{da:,.2f}")
print(f"PF (3%): ₹{pf:,.2f}")
print(f"Taxable Salary: ₹{taxable_salary:,.2f}")
print(f"Tax Rate: {tax_percent}%")
print(f"Tax Amount: ₹{tax_amount:,.2f}")
print(f"Total Deductions: ₹{total_deductions:,.2f}")
print("-"*50)
print(f"In-Hand Monthly Salary: ₹{(in_hand_salary/12):,.2f}")
print(f"In-Hand Annual Salary: ₹{in_hand_salary:,.2f}")
print("="*50)
````

### `Problem 2`: Write a program that take a user input of three angles and will find out whether it can form a triangle or not.

````python
# Write code here
# Take input from user
angle1 = float(input("Enter first angle: "))
angle2 = float(input("Enter second angle: "))
angle3 = float(input("Enter third angle: "))

# Calculate sum of angles
sum_of_angles = angle1 + angle2 + angle3

# Check if it can form a triangle
if angle1 > 0 and angle2 > 0 and angle3 > 0:
    if sum_of_angles == 180:
        print("These angles CAN form a triangle!")
        print(f"Sum of angles = {sum_of_angles}°")
    else:
        print("These angles CANNOT form a triangle.")
        print(f"Sum of angles = {sum_of_angles}° (should be 180°)")
else:
    print("Invalid angles! All angles must be greater than 0.")
    print(f"Angles entered: {angle1}°, {angle2}°, {angle3}°")
````

### `Problem 3`: Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit.

````python
# Write code here

cost_price = float(input("Enter Cost Price (CP): "))
selling_price = float(input("Enter Selling Price (SP): "))

# Calculate difference
difference = selling_price - cost_price

# Check profit or loss
if selling_price > cost_price:
    print("\n PROFIT!")
    print(f"Profit = ₹{difference:.2f}")
    profit_percent = (difference / cost_price) * 100
    print(f"Profit Percentage = {profit_percent:.2f}%")

elif selling_price < cost_price:
    print("\n LOSS!")
    loss = cost_price - selling_price
    print(f"Loss = ₹{loss:.2f}")
    loss_percent = (loss / cost_price) * 100
    print(f"Loss Percentage = {loss_percent:.2f}%")

else:
    print("\n NO PROFIT, NO LOSS")
    print("Selling Price = Cost Price (Break-even)")
````

### `Problem 5` - Exercise 12: Display Fibonacci series up to 10 terms.

*Note: The Fibonacci Sequence is a series of numbers. The next number is found by adding up the two numbers before it. The first two numbers are 0 and 1. For example, 0, 1, 1, 2, 3, 5, 8, 13, 21. The next number in this series above is 13+21 = 34*

````python
# Write code here
# Display Fibonacci series up to 10 terms

# Initialize the first two numbers
first = 0
second = 1

print("Fibonacci Series up to 10 terms:")

# Print the first two numbers
print(first)
print(second)

# Loop to generate next 8 terms (total 10 terms)
count = 2  # We already printed 2 terms
while count < 10:
    next_term = first + second  # Add the previous two numbers

    # Print the next term
    print(next_term)

    # Update values: shift to the next pair
    first = second
    second = next_term

    # Increment counter
    count = count + 1
````

### `Problem 7` - Reverse a given integer number.

Example:

`Input:`

```bash
76542
```

`Output:`

```bash
24567
```

````python
# Write code here
# Reverse a given integer number

num = int(input("Enter an integer to reverse: "))
original_num = num  # Store original for display
reverse = 0

# Handle negative numbers
is_negative = False
if num < 0:
    is_negative = True
    num = -num  # Make positive for processing

# Reverse the number
while num > 0:
    last_digit = num % 10  # Get last digit
    reverse = (reverse * 10) + last_digit  # Build reversed number
    num = num // 10  # Remove last digit

# If original was negative, make result negative
if is_negative:
    reverse = -reverse

print(f"Original number: {original_num}")
print(f"Reversed number: {reverse}")
````

### `Problem 8`: Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

**Example 1:**

`Input:`

```bash
30
```

`Output:`

```bash
276
```

````python
# Write code here
# Sum from 1 to N, skip numbers divisible by 5, stop if sum > 300

N = int(input())

total = 0
i = 1

while i <= N:
    # Skip numbers divisible by 5
    if i % 5 == 0:
        i = i + 1
        continue

    # Check if adding this number would exceed 300
    if total + i > 300:
        break

    # Add to sum
    total = total + i
    i = i + 1

print(total)
````

### `Problem 9`: Write a program that keeps on accepting a number from the user until the user enters Zero. Display the sum and average of all the numbers.

````python
# Write code here

total = 0
count = 0

print("Enter numbers (Enter 0 to stop):")

while True:
    num = int(input("Enter a number: "))

    # Check if user entered Zero
    if num == 0:
        break

    # Add to total and increment count
    total = total + num
    count = count + 1

# Display results
print("\n" + "=" * 30)
print(f"Sum of all numbers: {total}")
if count > 0:
    average = total / count
    print(f"Average of all numbers: {average}")
else:
    print("No numbers entered!")
````

###`Problem 9`: Write a program which will find all such numbers which are divisible by 7 but are not a multiple of 5, between 2000 and 3200 (both included). The numbers obtained should be printed in a comma-separated sequence on a single line.

````python
# Write code here


result = []

for num in range(2000, 3201):  # 3201 to include 3200
    if num % 7 == 0 and num % 5 != 0:
        result.append(str(num))

# Print as comma-separated sequence
print(",".join(result))
````

###`Problem 10`: Write a program, which will find all such numbers between 1000 and 3000 (both included) such that each digit of the number is an even number. The numbers obtained should be printed in a space-separated sequence on a single line.



````python
# Write code here
result = []

for num in range(1000, 3001):  # 3001 because 3000 is included
    # Convert number to string to check each digit
    num_str = str(num)
    all_even = True

    for digit in num_str:
        if int(digit) % 2 != 0:  # If any digit is odd
            all_even = False
            break

    if all_even:
        result.append(str(num))


print(" ".join(result))
````

###`Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps.
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps.

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.**

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python
import math

x = 0
y = 0

while True:
    command = input("Enter movement (or ! to stop): ")

    if command == "!":
        break

    direction, steps = command.split()
    steps = int(steps)

    if direction == "UP":
        y += steps
    elif direction == "DOWN":
        y -= steps
    elif direction == "RIGHT":
        x += steps
    elif direction == "LEFT":
        x -= steps

distance = math.sqrt(x**2 + y**2)
result = round(distance)
print(result)
````

###`Problem 12`:Write a program to print whether a given number is a prime number or not

````python
# Write code here
num = int(input("Enter a number: "))

# Prime numbers are greater than 1
if num <= 1:
    print("Not Prime")
else:
    is_prime = True
    # Check from 2 to num-1
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime")
    else:
        print("Not Prime")
````

###`Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
# Write code here
start = int(input("Enter start: "))
end = int(input("Enter end: "))

print(f"Armstrong numbers between {start} and {end}:")

for num in range(start, end + 1):
    temp = num
    sum_of_cubes = 0

    # Extract digits using while loop
    while temp > 0:
        digit = temp % 10  # Get last digit
        sum_of_cubes += digit ** 3
        temp = temp // 10   # Remove last digit

    if sum_of_cubes == num:
        print(num)
````

###`Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
# Write code here
H = int(input("Enter hours: "))
M = int(input("Enter minutes: "))

# Calculate angles
minute_angle = M * 6  # 6 degrees per minute
hour_angle = (H % 12) * 30 + M * 0.5  # 30 degrees per hour + 0.5 per minute

# Find the absolute difference
angle = abs(hour_angle - minute_angle)

# Get the minimum angle (there are two angles)
min_angle = min(angle, 360 - angle)

# Floor the result
result = int(min_angle)  # floor using int()

print(result)
````
