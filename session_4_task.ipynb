###`Problem 1:` Combine two lists index-wise(columns wise)

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

`Given List:`
```
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
```

`Output:`
```
[['M','y'], ['na', me'], ['i', 's'], ['Kh', 'an']]
```

````python
# Write code here
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]

result = [list(pair) for pair in zip(list1, list2)]
print(result)
````

###`Problem 3:` Update no of items available

Suppose you are given a list of candy and another list of same size representing no of items of respective candy.

i.e -  
```
candy_list = ['Jelly Belly','Kit Kat','Double Bubble','Milky Way','Three Musketeers']
no_of_items = [10,20,34,74,32]
```

Write a program to show no. of items of each candy type.

`Output:`

```
Jelly Belly-10
Kit Kat-20
Double Bubble-34
Milky Way-74
Three Musketeers-32

```

````python
# Write code here
candy_list = ['Jelly Belly', 'Kit Kat', 'Double Bubble', 'Milky Way', 'Three Musketeers']
no_of_items = [10, 20, 34, 74, 32]


for candy, count in zip(candy_list, no_of_items):
    print(f"{candy}-{count}")
````

###`Problem 4:` Running Sum on list
Write a program to print a list after performing running sum on it.

i.e:

`Input:`
```
list1 = [1,2,3,4,5,6]
```
`Output:`
```
[1,3,6,10,15,21]
```

````python
# Write code here
list1 = [1, 2, 3, 4, 5, 6]

result = []
running_sum = 0
for num in list1:
    running_sum += num
    result.append(running_sum)

print(result)
````

###`Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`




````python
# Write code here
numbers = [2, 4, 6, 10, 1]

result = [sum([num for num in numbers if num >= x]) for x in numbers]
print(result)
````

###`Problem 6:` Find list of common unique items from two list. and show in increasing order

`Input`

```
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

`Output:`
```
[34, 67, 89]
```


````python
# Write code here
num1 = [23, 45, 67, 78, 89, 34]
num2 = [34, 89, 55, 56, 39, 67]


common = [item for item in num1 if item in num2]

result = sorted(set(common))
print(result)
````

### `Problem 9:` Convert Character Matrix to single String using string comprehension.

**Example 1:**

Input:
```bash
[['c', 'a', 'm', 'p', 'u', 'x'], ['i', 's'], ['b', 'e', 's', 't'], ['c', 'h', 'a', 'n', 'n', 'e', 'l']]
```

Output:
```bash
campux is best channel
```

````python
# Write code here
matrix = [
    ['c', 'a', 'm', 'p', 'u', 'x'],
    ['i', 's'],
    ['b', 'e', 's', 't'],
    ['c', 'h', 'a', 'n', 'n', 'e', 'l']
]

# Convert character matrix to single string
result = ' '.join(''.join(row) for row in matrix)
print(result)
````

### `Problem 10:` Add Space between Potential Words.

**Example:**

Input:

```bash
['campusxIs', 'bestFor', 'dataScientist']
```

Output:
```bash
['campusx Is', 'best For', 'data Scientist']
```

````python
# Code here
words = ['campusxIs', 'bestFor', 'dataScientist']

result = [''.join([' ' + char if char.isupper() else char for char in word]) for word in words]
print(result)
````

### `Problem 11:` Write a program that can perform union operation on 2 lists

**Example:**

Input:

```bash
[1,2,3,4,5,1]
[2,3,5,7,8]
```

Output:
```bash
[1,2,3,4,5,7,8]
```

````python
# Write code here
list1 = [1, 2, 3, 4, 5, 1]
list2 = [2, 3, 5, 7, 8]

# Add elements from list2 that aren't in list1
union = list1 + [element for element in list2 if element not in list1]
print(union)  # [1, 2, 3, 4, 5, 1, 7, 8]  # Duplicate 1 still exists!

# To remove duplicates from list1 too
union = []
[union.append(x) for x in list1 + list2 if x not in union]
print(union)
````

### `Problem 14:` Write a list comprehension that can transpose a given matrix

matrix = [<br>
  [1,2,3],<br>
          [4,5,6],<br>
          [7,8,9]<br>]<br>

[1, 4, 7]<br>
[2, 5, 8]<br>
[3, 6, 9]<br>

````python
# Write code here
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Transpose using list comprehension
transpose = [[matrix[row][col] for row in range(len(matrix))] for col in range(len(matrix[0]))]

# Print the result
for row in transpose:
    print(row)
````
