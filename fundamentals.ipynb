# Type Conversion

```
#Implicit Vs Explicit
print(5+5.6)
print(type(5), type(5.6))

#Can't concatenate str and int
print(4+'4')
```

```
# Implicit conversion happens automatically by Python without any user intervention
# Python always converts smaller data types to larger data types (to avoid losing data)
# Order of promotion: int → float → complex (upcasting)


# Explicit
# Explicit conversion is done manually by the programmer using built-in Python functions. This is also called typecasting

# int(), str(), bool(), float(), list(), tuple(), dict(), set(), complex(), ord() - ASCII VALUE('A' -> 65) chr() - char from ASCII hex(), oct(), bin() - hex, oct, bin string

# str -> int
type(int('4'))
print(int(4.5))

# int(4+5j) # can't convert int to complex type

print(str(5))
print(float(4))

# Example 1: int + float = float (automatically converted)
num_int = 10          # int
num_float = 2.5       # float
result = num_int + num_float  # Python converts int to float
print(result)


# Cannot convert invalid string to int
# int("10.5")  # ValueError: invalid literal for int() with base 10: '10.5'
# first convert to float and then int
print(int(float('10.5')))

# Cannot convert non-numeric string to int
# int("Hello")  # ValueError: invalid literal for int() with base 10: 'Hello'

#  Cannot convert complex to int
# int(3+4j)     # TypeError: can't convert complex to int

# python doesn't change original value while conversion, its creates new one
```
