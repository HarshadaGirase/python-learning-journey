#Data Types
- Integer - can handle 1e309(1*10^309)
- Decimal/float - 1.7e309
- Boolean - True, False
- Text/String - Immutable sequence of Unicode characters
- Complex - Numbers with real and imaginary parts
- List - Mutable, ordered collection of items
- Tuple - Immutable, ordered collection of items
- Sets - Mutable, unordered collection of unique hashable objects
- Dictionary - Mutable, unordered (Python 3.7+: insertion ordered) collection of key-value pairs
- range - Immutable sequence of numbers (lazy evaluation)
- frozenset - Immutable version of set
