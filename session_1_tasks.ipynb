### Q2:- Write a program that will convert celsius value to fahrenheit.

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
import math
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))
# Calculate
distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)


print(f"Distance: {distance}")
````

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
# Write your code here

a = int(input("Enter first term: "))
b = int(input("Enter second term: "))
n = int(input("Enter N: "))

# Calculate
d = b - a                    # Common difference
nth_term = a + (n - 1) * d   # Nth term


print(nth_term)
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads = int(input("heads"))
legs= int(input("legs"))

extra_legs = legs - 2 * heads
dogs = extra_legs // 2
chickens = heads - dogs
print("Dogs", dogs)
print("Chickens:", chickens)
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here
no = int(input("Enter number: "))
total = 0

for i in range(1, no + 1):
    sq = i * i
    total += sq

print("Sum of squares:", total)
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
# Write your code here
n1 = int(input("1st Numerator: "))
d1 = int(input("1st Denominator: "))
n2 = int(input("2nd Numerator: "))
d2 = int(input("2nd Denominator: "))

# Calculate fraction
result_num = (n1 * d2) + (n2 * d1)
result_den = d1 * d2

# Show as fraction
print(f"Sum = {result_num}/{result_den}")
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
import math

# Tank dimensions
H = float(input("Enter tank height (cm): "))
L = float(input("Enter tank length (cm): "))
B = float(input("Enter tank breadth (cm): "))

# Glass dimensions
r = float(input("Enter glass radius (cm): "))
h = float(input("Enter glass height (cm): "))

# Calculate volumes
tank_volume = H * L * B  # cm³
glass_volume = math.pi * r * r * h  # cm³

# Calculate number of glasses
num_glasses = tank_volume // glass_volume  # Full glasses only


print(f"Tank volume: {tank_volume:.2f} cm³")
print(f"Glass volume: {glass_volume:.2f} cm³")
print(f"Number of full glasses: {int(num_glasses)}")
````
